```python
import pandas as pd

iris = pd.read_csv('/kaggle/input/datasets/organizations/uciml/iris/Iris.csv')

iris.head()
```

```python
iris.info()
```

```python
import matplotlib.pyplot as plt

plt.scatter(iris['SepalLengthCm'], iris['SepalWidthCm'], c=iris['Species'].astype('category').cat.codes)

plt.title("Sepal Length vs Sepal Width")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")

plt.show()
```

## Scatter Plot: Sepal Length vs Sepal Width

This scatter plot visualizes the relationship between Sepal Length and Sepal Width in the Iris dataset.

Each point represents a flower sample, and colors represent different species:
- Setosa
- Versicolor
- Virginica

### Observation:
- Setosa forms a clearly separable cluster.
- Versicolor and Virginica show partial overlap.
- This indicates that multiple features are required for accurate classification.

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))
sns.heatmap(iris.drop(columns=['Id', 'Species']).corr(), annot=True)

plt.title("Feature Correlation Heatmap")
plt.show()
```

## Correlation Heatmap of Iris Features

This heatmap shows the correlation between numerical features in the Iris dataset.

Correlation values range from:
- +1 → Strong positive relationship
- 0 → No relationship
- -1 → Strong negative relationship

### Observation:
- Petal Length and Petal Width show strong positive correlation.
- Sepal features have weaker correlation compared to petal features.
- This suggests petal measurements are more important for classification.

```python
sns.pairplot(iris.drop(columns=['Id']), hue='Species')

plt.show()
```

## Pairplot Analysis of Iris Dataset

This pairplot visualizes relationships between all numerical features in the Iris dataset.

Each plot shows a pairwise relationship between two features, and colors represent different species:
- Setosa
- Versicolor
- Virginica

### Observation:
- Setosa is clearly separable from the other two species in most feature combinations.
- Versicolor and Virginica show some overlap, especially in sepal-based features.
- Petal-based features (Petal Length and Petal Width) provide better separation between species.

### Insight:
This indicates that petal measurements are more effective for classification compared to sepal measurements.
