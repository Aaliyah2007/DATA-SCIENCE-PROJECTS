```python
import pandas as pd

netflix = pd.read_csv('/kaggle/input/datasets/shivamb/netflix-shows/netflix_titles.csv')

netflix.head()
```

```python
netflix.info()
```

```python
import matplotlib.pyplot as plt

netflix['type'].value_counts().plot(kind='bar')

plt.title("Movies vs TV Shows on Netflix")
plt.xlabel("Content Type")
plt.ylabel("Count")

plt.show()
```

## Movies vs TV Shows Distribution

This visualization compares the number of Movies and TV Shows available in the Netflix dataset.

### Observation
- Movies are more common than TV Shows.
- Netflix's content library is dominated by movies.
- Content type distribution helps understand the platform's catalog strategy.

```python
import matplotlib.pyplot as plt

netflix['release_year'].value_counts().sort_index().plot()

plt.title("Netflix Content Released by Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.show()
```

## Netflix Content Release Trend

This visualization shows the number of titles released each year in the Netflix dataset.

### Observation
- The number of releases increases significantly in recent years.
- Netflix has expanded its content library rapidly.
- The trend indicates growing investment in content production and acquisition.

```python
import matplotlib.pyplot as plt

netflix['country'].dropna().value_counts().head(10).plot(kind='bar')

plt.title("Top 10 Countries by Netflix Content")
plt.xlabel("Country")
plt.ylabel("Number of Titles")

plt.show()
```

## Top Countries Producing Netflix Content

This visualization shows the countries with the highest number of titles in the Netflix dataset.

### Observation
- Some countries contribute significantly more content than others.
- The United States is typically among the largest contributors.
- Content production is concentrated in a few major entertainment markets.

### Insight
Country-wise analysis helps understand Netflix's global content distribution and regional influence.

```python
import matplotlib.pyplot as plt

netflix['listed_in'].value_counts().head(10).plot(kind='bar')

plt.title("Top 10 Netflix Genres")
plt.xlabel("Genre")
plt.ylabel("Number of Titles")

plt.show()
```

## Most Popular Genres on Netflix

This visualization shows the most common content categories in the Netflix dataset.

### Observation
- Certain genres dominate Netflix's catalog.
- Drama, Comedy, and Documentary-related categories often appear among the most common genres.
- Genre analysis helps understand audience preferences and content strategy.

### Insight
Understanding genre distribution provides insight into the types of content Netflix prioritizes for its viewers.
