```python
import pandas as pd

train = pd.read_csv('/kaggle/input/competitions/titanic/train.csv')
train.head()
```

```python
train.info()
```

```python
import matplotlib.pyplot as plt

train ['Survived'].value_counts().plot(kind='bar')

plt.title("Survival Count (0 = Not Survived, 1 = Survived)")

plt.show()
```

```python
import matplotlib.pyplot as plt

gender_survival = train.groupby('Sex')['Survived'].mean()

gender_survival.plot(kind='bar')

plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Survival Rate")

plt.show()
```

## Observation

Female passengers had a significantly higher survival rate compared to male passengers.
This suggests gender strongly influenced survival during the Titanic disaster.
